```python
def get_area(centerx = 640,centrey = 512):
    canvas = np.zeros(shape=fs.slm.shape)
    # Imprint three windows, with different blazes.
    x = 64
    dk = .004
    toolbox.imprint(canvas, window=[ centerx-x, 2*x, centrey-x, 2*x], function=phase.blaze, grid=fs.slm, vector=( -dk,  dk))
    img = plot_cam(fs=fs,phase=canvas)
    print(img[150:350,900:1100].sum())
```
