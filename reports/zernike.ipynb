```python
subset = pdf[
(pdf['zernike_index'] == 16) &
(pdf['expected_y'] == 380) &
(pdf['expected_x'] == 1030)
]

ax = subset.plot(x="w", y="r_squared", label="r_squared", color="blue")
subset.plot(x="w", y="strehl_prime", label="strehl_prime", color="red", ax=ax, secondary_y=True)
```
